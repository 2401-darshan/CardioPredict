### 1. Handle missing values

### Step 1.1 Check for missing values in each column

```python
print('Missing values per column:')
print(cardio.isnull().sum())
print('\nTotal missing values:', cardio.isnull().sum().sum())
```

### Step 1.2 conclusion on missing values

```python
if cardio.isnull().sum().sum() == 0:
    print('No missing values found in the dataset!')
else:
    num_cols = cardio.select_dtypes(include=np.number).columns
    cardio[num_cols] = cardio[num_cols].fillna(cardio[num_cols].median())
    print('Missing values filled with median for numerical columns.')
```

---
## 2. Identify and Handle Outliers

### Step 2.1 Check statistical summary to spot outliers

```python
cardio.describe()
```

### Step 2.2 Visualize outliers using Boxplots

```python
columns = ['height', 'weight', 'ap_hi', 'ap_lo']

fig, axes = plt.subplots(1, 4, figsize=(14, 5))
for i, col in enumerate(columns):
    axes[i].boxplot(cardio[col])
    axes[i].set_title(col)
    axes[i].set_ylabel('Value')

plt.suptitle('Boxplots to Identify Outliers', fontsize=15)
plt.show()
```

### Step 2.3 Identify outliers using IQR method

```python
for col in columns:
    Q1 = cardio[col].quantile(0.25)
    Q3 = cardio[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    outlier_count = cardio[(cardio[col] < lower) | (cardio[col] > upper)].shape[0]
    print(f'{col}: Lower={lower:.1f}, Upper={upper:.1f}, Outliers={outlier_count}')
```

### Step 2.4 Remove outliers using IQR (Filter unrealistic medical values)

```python
print('Shape before removing outliers:', cardio.shape)

# Removes medically impossible values
cardio = cardio[
    (cardio['ap_hi'] >= 80)  & (cardio['ap_hi'] <= 250) &   # systolic BP
    (cardio['ap_lo'] >= 50)  & (cardio['ap_lo'] <= 180) &   # diastolic BP
    (cardio['height'] >= 100) & (cardio['height'] <= 220) &  # height in cm
    (cardio['weight'] >= 30)  & (cardio['weight'] <= 200)    # weight in kg
]

print('Shape after removing outliers :', cardio.shape)
print('Rows removed:', 70000 - cardio.shape[0])
```

### Model creation without using Library

```python
import pandas as pd
import numpy as np

df = pd.read_csv('iphone.csv')
X = df['number'].values
Y = df['Price'].values
n = len(X)
m, c = 0.0, 0.0 

lr = 0.001 
epochs = 300
loss_history = []

for i in range(epochs):
    Y_pred = m * X + c                     
    error = Y_pred - Y                    
    loss = (1/n) * np.sum(error ** 2)      
    loss_history.append(loss)
    
    dm = (2/n) * np.sum(error * X)
    dc = (2/n) * np.sum(error)            
    
    m = m - lr * dm                        
    c = c - lr * dc                      
    
    if i % 4000 == 0:
        print('epochs: ', i)
        print('loss: ', loss)
        print('m: ', m)
        print('c: ', c)
print('\nFinal parameters : m:', m, ' c:', c)
x = 18
print("Y = ",m*x+c)
```

### Plot in Graph

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(8, 5))
plt.plot(range(1, epochs + 1), loss_history, color='indigo', linewidth=2)
plt.title('Epochs vs Loss Function (MSE) - iPhone Dataset', fontsize=14, fontweight='bold')
plt.xlabel('Epochs', fontsize=12)
plt.ylabel('Loss Mean Squared Error', fontsize=12)
plt.show()
```

## import libraries

```python
import pandas as pd

from sklearn.model_selection import train_test_split, cross_val_score, GridSearchCV

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
```

## 1. Load Data


```python
# Load the dataset
cardio = pd.read_csv("cardio_train_official.csv", sep=';')

print("Dataset loaded successfully!")
print("Dataset Shape:", cardio.shape)
```

## 2 - PREPARE FEATURES AND TARGET

```python
# Target column
y = cardio["cardio"]

# Features
X = cardio.drop("cardio", axis=1, inplace=False)

X
```

### 3. TRAIN TEST SPLIT

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("\nTraining data:", X_train.shape)
print("Testing data :", X_test.shape)
```

### 4 - CREATE CLASSIFICATION MODELS

```python
from sklearn.preprocessing import StandardScaler

# Scale features for Logistic Regression
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)


logistic_model = LogisticRegression(max_iter=1000)

decision_tree_model = DecisionTreeClassifier(
    random_state=42
)

random_forest_model = RandomForestClassifier(
    n_estimators=50,
    random_state=42,
    n_jobs=-1
)

models = {
    "Logistic Regression": logistic_model,
    "Decision Tree": decision_tree_model,
    "Random Forest": random_forest_model
}
```

### 5 - MODEL EVALUATION

```python
results = []

for name, model in models.items():

    print("\nRunning:", name)

    # Logistic Regression uses scaled data
    if name == "Logistic Regression":
        train_data = X_train_scaled
        test_data = X_test_scaled
    else:
        train_data = X_train
        test_data = X_test

    # Train model
    model.fit(train_data, y_train)

    # Predictions
    y_train_pred = model.predict(train_data)
    y_test_pred = model.predict(test_data)

    # Metrics
    accuracy = accuracy_score(y_test, y_test_pred)

    precision = precision_score(
        y_test,
        y_test_pred,
        zero_division=0
    )

    recall = recall_score(
        y_test,
        y_test_pred,
        zero_division=0
    )

    f1 = f1_score(
        y_test,
        y_test_pred,
        zero_division=0
    )

    # Train/Test accuracy
    train_accuracy = accuracy_score(
        y_train,
        y_train_pred
    )

    test_accuracy = accuracy_score(
        y_test,
        y_test_pred
    )

    # 5-Fold Cross Validation
    cv_scores = cross_val_score(
        model,
        train_data,
        y_train,
        cv=5,
        scoring="accuracy",
        n_jobs=-1
    )

    cv_average = cv_scores.mean()
    cv_spread = cv_scores.max() - cv_scores.min()

    results.append([
        name,
        accuracy,
        precision,
        recall,
        f1,
        train_accuracy,
        test_accuracy,
        cv_average,
        cv_spread
    ])

    print("Completed:", name)
```

### 6 - MODEL COMPARISON

```python
comparison_df = pd.DataFrame(
    results,
    columns=[
        "Model",
        "Accuracy",
        "Precision",
        "Recall",
        "F1-score",
        "Train Accuracy",
        "Test Accuracy",
        "CV Average",
        "CV Spread"
    ]
)

print("\n\n================================================")
print("MODEL COMPARISON")
print("================================================")

print(comparison_df.round(4))
```

### 7 - OVERFITTING / UNDERFITTING

```python
print("OVERFITTING / UNDERFITTING CHECK")

for row in results:

    model_name = row[0]
    train_accuracy = row[5]
    test_accuracy = row[6]

    difference = train_accuracy - test_accuracy

    print("\nModel:", model_name)
    print("Training Accuracy:", round(train_accuracy, 4))
    print("Testing Accuracy :", round(test_accuracy, 4))
    print("Difference       :", round(difference, 4))

    if difference > 0.10:
        print("Result: Possible Overfitting")

    elif train_accuracy < 0.70 and test_accuracy < 0.70:
        print("Result: Possible Underfitting")

    else:
        print("Result: Good Fit")
```

### 8 --> 5-FOLD CROSS VALIDATION

```python
print("5-FOLD CROSS VALIDATION...")

for row in results:

    model_name = row[0]
    cv_average = row[7]
    cv_spread = row[8]

    print("\nModel:", model_name)
    print("Average CV Score:", round(cv_average, 4))
    print("CV Score Spread :", round(cv_spread, 4))
```

### 9 - RANDOM FOREST HYPERPARAMETER TUNING

```python
print("RANDOM FOREST HYPERPARAMETER TUNING")

rf = RandomForestClassifier(
    random_state=42
)

# Parameters to test
param_grid = {
    "n_estimators": [50, 100],
    "max_depth": [None, 10, 20]
}


# GridSearchCV
grid_search = GridSearchCV(
    rf,
    param_grid,
    cv=5,
    scoring="accuracy"
)


# Train GridSearchCV
grid_search.fit(X_train, y_train)


# Best parameters
print("Best Parameters:", grid_search.best_params_)

# Best CV score
print(
    "Best CV Score:",
    round(grid_search.best_score_, 4)
)
```

### 10. TEST TUNED RANDOM FOREST

```python
best_rf = grid_search.best_estimator_

# Predictions
tuned_pred = best_rf.predict(X_test)


# Calculate final metrics
tuned_accuracy = accuracy_score(
    y_test, tuned_pred
)

tuned_precision = precision_score(
    y_test,
    tuned_pred,
    zero_division=0
)

tuned_recall = recall_score(
    y_test,
    tuned_pred,
    zero_division=0
)

tuned_f1 = f1_score(
    y_test,
    tuned_pred,
    zero_division=0
)


print("\n\n================================================")
print("TUNED RANDOM FOREST RESULTS")
print("================================================")

print("Accuracy :", round(tuned_accuracy, 4))
print("Precision:", round(tuned_precision, 4))
print("Recall   :", round(tuned_recall, 4))
print("F1-score :", round(tuned_f1, 4))
```
